```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import cm

def anomaly_wave(x, y):
    decay_factor = np.exp(0.15 * np.sqrt(x**2 + y**2))
    return (np.sin(np.sqrt(x**2 + y**2)) * np.cos(x)) / decay_factor

x = np.linspace(-12, 12, 500)
y = np.linspace(-12, 12, 500)
X, Y = np.meshgrid(x, y)
Z = anomaly_wave(X, Y)

fig = plt.figure(figsize=(12, 7), facecolor='#0d1117')
ax = fig.add_subplot(111, projection='3d')
ax.set_facecolor('#0d1117')
surf = ax.plot_surface(X, Y, Z, cmap=cm.magma, linewidth=0.1, 
                       antialiased=True, edgecolor='#000000', alpha=0.9)

ax.set_title('MESSAGE ETHERNAL INTERCEPTION MATRIX\n[Live Parameter Parsing]', 
             color='#00ff00', pad=25, fontsize=13, fontweight='bold')
ax.set_xlabel('Encrypted Node X', color='#8b949e', labelpad=15)
ax.set_ylabel('Encrypted Node Y', color='#8b949e', labelpad=15)
ax.set_zlabel('Threat Level Variance', color='#8b949e', labelpad=15)


ax.xaxis.pane.fill = False
ax.yaxis.pane.fill = False
ax.zaxis.pane.fill = False
ax.grid(color='#30363d', linestyle=':', linewidth=0.8)


ax.set_xticklabels([])
ax.set_yticklabels([])
ax.set_zticklabels([])

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec

plt.style.use('dark_background')
np.random.seed(42)
data_x = np.random.normal(loc=50, scale=15, size=2000)
data_y = data_x * 1.2 + np.random.normal(loc=0, scale=15, size=2000)
fig = plt.figure(figsize=(14, 9), facecolor='#0d1117')
gs = gridspec.GridSpec(2, 2, figure=fig, height_ratios=[1.2, 1])

ax1 = fig.add_subplot(gs[0, :])
ax1.set_facecolor('#0d1117')
count, bins, ignored = ax1.hist(data_x, 40, density=True, color='#8a2be2', alpha=0.5, edgecolor='#0d1117')

bell_curve = 1/(15 * np.sqrt(2 * np.pi)) * np.exp( - (bins - 50)**2 / (2 * 15**2) )
ax1.plot(bins, bell_curve, linewidth=2.5, color='#00ff00')
ax1.set_title('GAUSSIAN DISTRIBUTION MODEL\n[Probability Density Function]', color='#00ff00', pad=15, fontweight='bold')
ax1.grid(color='#30363d', linestyle='--', linewidth=0.5, alpha=0.7)

ax2 = fig.add_subplot(gs[1, 0])
ax2.set_facecolor('#0d1117')
ax2.scatter(data_x, data_y, color='#00d2ff', alpha=0.4, s=15)

m, b = np.polyfit(data_x, data_y, 1)
ax2.plot(data_x, m*data_x + b, color='#ff007f', linewidth=2.5)
ax2.set_title('LINEAR REGRESSION VARIANCE', color='#00d2ff', pad=10, fontweight='bold')
ax2.grid(color='#30363d', linestyle=':', linewidth=0.8)

ax3 = fig.add_subplot(gs[1, 1])
ax3.set_facecolor('#0d1117')
hb = ax3.hexbin(data_x, data_y, gridsize=25, cmap='inferno', edgecolors='#0d1117', linewidths=0.5)
ax3.set_title('BIVARIATE HEXAGONAL DENSITY', color='#ffaa00', pad=10, fontweight='bold')
plt.tight_layout(pad=3.0)

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.gridspec as gridspec

plt.style.use('dark_background')
fig = plt.figure(figsize=(16, 9), facecolor='#11151c')
gs = gridspec.GridSpec(2, 3, figure=fig, wspace=0.25, hspace=0.35)
TXT_CLR = '#94a3b8'
GRID_CLR = '#1e293b'
BLUE_LINE = '#38bdf8'
AMBER_LINE = '#fbbf24'
ax1 = fig.add_subplot(gs[0, 0:2], facecolor='#11151c')
t = np.linspace(0, 10, 5000)

signal = 2.5 * np.sin(2 * np.pi * 1.5 * t) + 0.5 * np.cos(2 * np.pi * 8 * t)
noise = np.random.normal(0, 0.2, len(t))
ax1.plot(t, signal + noise, color='#334155', linewidth=0.8, label='Raw Input Sensor')
ax1.plot(t, signal, color=BLUE_LINE, linewidth=1.2, label='Filtered Output (RX-28)')
ax1.set_title('HIGH-RESOLUTION TELEMETRY STREAM', color=TXT_CLR, fontsize=10, fontweight='bold', pad=10)
ax1.set_ylabel('Amplitude (mV)', color=TXT_CLR, fontsize=9)
ax1.grid(color=GRID_CLR, linestyle='-', linewidth=0.8)
ax1.legend(loc='upper right', frameon=False, labelcolor=TXT_CLR, fontsize=8)
ax2 = fig.add_subplot(gs[0, 2], facecolor='#11151c')
t_phase = np.linspace(0, 2*np.pi, 2000)


x_phase = np.sin(5 * t_phase + np.pi/2) * np.exp(-0.1 * t_phase)
y_phase = np.sin(4 * t_phase) * np.exp(-0.1 * t_phase)
ax2.plot(x_phase, y_phase, color=AMBER_LINE, linewidth=0.6)
ax2.set_title('PHASE MARGIN TRAJECTORY', color=TXT_CLR, fontsize=10, fontweight='bold', pad=10)
ax2.grid(color=GRID_CLR, linestyle='--', linewidth=0.5)
ax3 = fig.add_subplot(gs[1, 0], facecolor='#11151c')
matrix_x = np.linspace(-3, 3, 150)
matrix_y = np.linspace(-3, 3, 150)
X, Y = np.meshgrid(matrix_x, matrix_y)
Z = np.exp(-(X**2 + Y**2)/2.0) * np.cos(4 * X) * np.sin(4 * Y)
cax = ax3.imshow(Z, cmap='cividis', aspect='auto', origin='lower', extent=[-3, 3, -3, 3])

ax3.set_title('SPATIAL FREQUENCY ANALIST DATA', color=TXT_CLR, fontsize=10, fontweight='bold', pad=10)
ax3.set_xlabel('Vector X', color=TXT_CLR, fontsize=9)
ax3.set_ylabel('Vector Y', color=TXT_CLR, fontsize=9)
ax4 = fig.add_subplot(gs[1, 1:3], facecolor='#11151c')

data_points = np.random.normal(15, 2.5, 10000)
ax4.hist(data_points, bins=80, color='#1e293b', edgecolor='#3b82f6', density=True, alpha=0.7)
x_curve = np.linspace(5, 25, 200)
y_curve = (1 / (2.5 * np.sqrt(2 * np.pi))) * np.exp(-0.5 * ((x_curve - 15) / 2.5)**2)
ax4.plot(x_curve, y_curve, color='#e2e8f0', linewidth=1.5, linestyle='--')
ax4.axvline(15, color='#ef4444', linewidth=1.2, linestyle='-', label='Mean (μ)')
ax4.axvline(15+2.5, color=AMBER_LINE, linewidth=1, linestyle=':', label='1σ (68.2%)')
ax4.axvline(15-2.5, color=AMBER_LINE, linewidth=1, linestyle=':')
ax4.set_title('LATENCY DISTRIBUTION & SYSTEM DEVIATION', color=TXT_CLR, fontsize=10, fontweight='bold', pad=10)
ax4.set_xlabel('Latency (ms)', color=TXT_CLR, fontsize=9)
ax4.grid(color=GRID_CLR, linestyle='-', linewidth=0.5)
ax4.legend(loc='upper right', frameon=False, labelcolor=TXT_CLR, fontsize=8)
    
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
np.random.seed(42) 
no_absen = np.arange(1, 31)
nilai_ujian = np.random.randint(65, 96, size=30)
nilai_ujian[27] = 99 
rata_rata = np.mean(nilai_ujian)
nilai_tertinggi = np.max(nilai_ujian)
nilai_terendah = np.min(nilai_ujian)
jumlah_lulus = np.sum(nilai_ujian >= 75) 

print(">>> LAPORAN ANALISIS DATA KELAS <<<")
print(f"Total Siswa      : {len(nilai_ujian)} orang")
print(f"Rata-rata Kelas  : {rata_rata:.2f}")
print(f"Nilai Tertinggi  : {nilai_tertinggi}")
print(f"Nilai Terendah   : {nilai_terendah}")
print(f"Siswa Lulus KKM  : {jumlah_lulus} orang ({(jumlah_lulus/30)*100:.1f}%)")
print("===================================")

plt.style.use('dark_background')
fig, ax = plt.subplots(figsize=(12, 5), facecolor='#0d1117')
ax.set_facecolor('#0d1117')
warna_bar = ["#ff0000" if nilai == nilai_tertinggi else '#38bdf8' for nilai in nilai_ujian]
bars = ax.bar(no_absen, nilai_ujian, color=warna_bar, edgecolor='#0d1117')
ax.axhline(rata_rata, color='#00ff00', linestyle='--', linewidth=1.5, label=f'Rata-rata ({rata_rata:.1f})')
ax.set_title('DISTRIBUSI NILAI PRAKTIK PEMROGRAMAN\n[SMKN 10 SURABAYA]', color='#00ff00', pad=15, fontweight='bold')
ax.set_xlabel('Nomor Absen', color='#94a3b8')
ax.set_ylabel('Skor Ujian', color='#94a3b8')
ax.set_xticks(no_absen)
ax.tick_params(axis='x', labelsize=8)
ax.grid(axis='y', color='#1f2937', linestyle=':', linewidth=1)
ax.legend(facecolor='#0d1117', edgecolor='#1f2937', labelcolor='#94a3b8')
plt.tight_layout()

plt.show()
```
